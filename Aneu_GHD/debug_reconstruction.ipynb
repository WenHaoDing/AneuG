```python
from utils.patching import planarize_openings, smooth_near_openings
import pyvista as pv
import os
import utils.vessel_reconstruct as vr

case = "72rwT0u5nJ_aneurysm1"
ghd_dir = "fitting_results/ImperialNHS_batch1"
post_dir = "seg_post/ImperialNHS_batch1"
mesh_path = os.path.join(ghd_dir, case, "ghd_fitted_uncapped_world.obj")
mesh = pv.read(mesh_path)
mesh_out = planarize_openings(
    mesh,
    post_dir=os.path.join(post_dir, case),
    r_forward_fusion_info_filename="forward_fusion_info.npz",
    n_iter=10,
    lam=0.5
)
pl = pv.Plotter()
pl.add_mesh(mesh_out, color="lightblue", opacity=0.8, show_edges=True)
pl.show()
pv.save_meshio(os.path.join(post_dir, case, "ghd_reconstructed.obj"), mesh_out)

# merge
w_merged_mesh_filename = "ghd_merged_reconstruction.ply"

vr.forward_mesh_fusion_v2(post_dir=os.path.join(post_dir, case),
                          r_clipped_mesh_filename="ghd_reconstructed.obj",
                          w_forward_fusion_info_filename="ghd_forward_fusion_info.npz",
                          w_merged_mesh_filename=w_merged_mesh_filename)
merged_mesh_path = os.path.join(post_dir, case, w_merged_mesh_filename)
merged_mesh = pv.read(merged_mesh_path)
pl = pv.Plotter()
pl.add_mesh(merged_mesh, color="lightblue", opacity=0.8, show_edges=True)
pl.show()

smoothed_mesh = smooth_near_openings(merged_mesh, os.path.join(post_dir, case), n_rings=3, n_iter=10, lam=0.5)
pl = pv.Plotter()
pl.add_mesh(smoothed_mesh, color="lightblue", opacity=0.8, show_edges=True)
pl.show()
```
