```python
import os
import shutil
import utils.vessel_reconstruct as vr


case = "euHpN3eKdM_aneurysm1"
ghd_root = "fitting_results/ImperialNHS_batch1"
seg_post_root = "seg_post"

ghd_dir = os.path.join(ghd_root, case)
seg_post_dir = os.path.join(seg_post_root, case)
if os.path.exists(ghd_dir) and os.path.exists(seg_post_dir):
    r_clipped_mesh_filename_src = os.path.join(ghd_dir, "ghd_fitted_uncapped_world.obj")
    r_clipped_mesh_filename = "ghd_fitted_uncapped_world.obj"
    shutil.copy2(r_clipped_mesh_filename_src, os.path.join(seg_post_dir, r_clipped_mesh_filename))
    r_clipped_cl_filename = "clipped_centerline.npy"
    w_merged_mesh_filename = "ghd_fitted_merged.obj"
    vr.forward_mesh_fusion_v2(post_dir=seg_post_dir,
                              flaw_opening_min_size=10,
                              init_step=4,
                              max_cl_length=10,
                              ring_downsample_ratio=1,
                              r_clipped_mesh_filename=r_clipped_mesh_filename,
                              r_clipped_cl_filename=r_clipped_cl_filename,
                              w_forward_fusion_info_filename=None,
                              w_merged_mesh_filename=w_merged_mesh_filename
    )
    
```

```python
import pyvista as pv

merged_mesh_path = os.path.join(seg_post_dir, w_merged_mesh_filename)
mesh = pv.read(merged_mesh_path)

pl = pv.Plotter()
pl.add_mesh(mesh, color="lightblue", opacity=0.8, show_edges=True)
pl.show()
```
