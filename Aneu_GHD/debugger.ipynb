```python
import os
import pyvista as pv

ghd_root = "fitting_results/ImperialNHS_batch1"
post_root = "/media/yaplab2/HDD Storage/wenhao/AneuSeg/PostTr_debug/ImperialNHS_batch1"
output_dir = "mesh_comparisons"
os.makedirs(output_dir, exist_ok=True)

case_list = os.listdir(ghd_root)
original_mesh_name_list = ["clipped_reconstruction_fallback.ply", "clipped_reconstruction.ply"]
fitted_mesh_name = "ghd_fitted_uncapped_world.obj"

for case in case_list:
    try:
        # Pick original mesh: use first name if both exist, otherwise whichever is available
        found = [n for n in original_mesh_name_list
                 if os.path.exists(os.path.join(post_root, case, n))]
        if not found:
            print(f"[{case}] No original mesh found, skipping.")
            continue
        original_mesh_path = os.path.join(post_root, case, found[0])

        fitted_mesh_path = os.path.join(ghd_root, case, fitted_mesh_name)
        if not os.path.exists(fitted_mesh_path):
            print(f"[{case}] Fitted mesh not found, skipping.")
            continue

        original_mesh = pv.read(original_mesh_path)
        fitted_mesh = pv.read(fitted_mesh_path)

        pl = pv.Plotter(off_screen=True, window_size=(1000, 1000))
        pl.add_mesh(original_mesh, color="blue", opacity=0.3, show_edges=False, label="Original")
        pl.add_mesh(fitted_mesh, color="red", opacity=0.3, show_edges=True, edge_color="black", label="Fitted")
        pl.add_legend(face=None)
        pl.add_title(case, font_size=10)
        pl.reset_camera()

        out_path = os.path.join(output_dir, f"{case}.png")
        pl.screenshot(out_path)
        pl.close()
        print(f"[{case}] Saved → {out_path}")

    except Exception as e:
        print(f"[{case}] Error: {e}")
```
